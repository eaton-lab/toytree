# Rooting on an outgroup

Manual rooting is the most direct rooting workflow in `toytree`. You choose the edge below the new root by calling `tree.root(...)`, and you can undo the root split later with `tree.unroot()`. This is the right method when you know which tip or clade should be the outgroup.


```python
import toytree
```

## Main functions

- `tree.root(*query, root_dist=None, edge_features=None)`
- `tree.unroot()`

The `query` selects the node below the edge that will be split by the new root. If you pass multiple selectors, `toytree` finds their MRCA and roots on the edge above that node.


```python
# rooted reference tree and an unrooted copy
base = toytree.tree("(((a:1,b:1):1,c:2):1,(d:1,e:1):2);")
utree = base.unroot()
```

## Root, unroot, and re-root

Here the clade `(d, e)` is the outgroup in the rooted reference tree. We can unroot the tree, recover the original rooting, and then re-root it on a different edge.


```python
rtree = utree.root("d", "e")
atree = rtree.root("c")

c, a, m = toytree.mtree([utree, rtree, atree]).draw(
    layout="d",
    node_labels="idx",
    node_sizes=14,
    tip_labels=True,
    width=700,
    height=250,
)
a[0].label.text = "unrooted"
a[1].label.text = "root on (d,e)"
a[2].label.text = "re-root on c"
c
```

## Selecting the outgroup

You can root on a single tip, on multiple tips that define a monophyletic clade, or on an internal node directly. The important requirement is that the selected outgroup is monophyletic in the current tree.


```python
# root on a single-tip outgroup
utree.root("e").draw(layout="d", node_labels="idx", node_sizes=14);
```

```python
# root on the MRCA of multiple tips
utree.root("d", "e").draw(layout="d", node_labels="idx", node_sizes=14);
```

```python
# root on an internal node selected directly
mrca = utree.get_mrca_node("d", "e")
utree.root(mrca).draw(layout="d", node_labels="idx", node_sizes=14);
```

## Controlling the root position with `root_dist`

Selecting the correct edge is only part of rooting. You also need a position on that edge. If `root_dist=None`, `tree.root()` places the root at the midpoint of the chosen edge. You can also set the distance from the selected node explicitly.


```python
# default midpoint placement
utree.root("d", "e").draw(layout="d", node_labels="idx", node_sizes=14);
```

```python
# move the root 0.25 units above the selected clade
utree.root("d", "e", root_dist=0.25).draw(layout="d", node_labels="idx", node_sizes=14);
```

## Preserving edge-associated data with `edge_features`

Branch lengths and support values are already treated as edge data during rooting. For any custom feature that really belongs to an edge, pass its name in `edge_features` so it is re-polarized correctly when the root moves.

The example below is adapted from the older rooting tutorial and the Czech et al. support-value example. Node colors and edge colors are stored separately so you can see what changes when the tree is re-rooted.


```python
czech = toytree.tree("((C,D)1,(A,(B,X)3)2,E)R;", internal_labels="name")
colors = {"1": "red", "2": "green", "3": "orange"}
ctree = czech.set_node_data("ecolor", colors, default="black")
ctree = ctree.set_node_data("ncolor", colors, default="black")

kwargs = {
    "layout": "d",
    "use_edge_lengths": False,
    "node_sizes": 10,
    "node_labels": "name",
    "node_labels_style": {
        "font-size": 20,
        "baseline-shift": 10,
        "-toyplot-anchor-shift": 10,
    },
}

ctree.draw(node_colors="ncolor", edge_colors="ecolor", **kwargs);
```

```python
# re-root while preserving 'ecolor' as edge-associated data
rtree = ctree.root("X", edge_features=["ecolor"])
rtree.draw(node_colors="ncolor", edge_colors="ecolor", **kwargs);
```
