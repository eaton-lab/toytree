# Rooting trees

Rooting or re-rooting a tree assigns direction to ancestor-descendant relationships. Most phylogenetic inference methods return an unrooted topology, so rooting is the step where you decide which edge represents the split between the outgroup and the ingroup, or where an algorithm infers that split from branch lengths or reconciliation information.

This overview keeps the conceptual material from the older rooting tutorial, but moves the method-specific workflows into three dedicated pages.


```python
import toytree
```

## A simple example

The example below shows the same topology in three states: rooted, unrooted, and rooted again on a different edge.


```python
# an example tree with outgroup (r3,r4)
tree = toytree.rtree.unittree(ntips=5, seed=123)
utree = tree.unroot()
rtree = utree.root("r3", "r4")
atree = rtree.root("r2")
```

```python
c, a, m = toytree.mtree([tree, utree, atree]).draw(
    layout="d",
    node_labels="idx",
    node_sizes=14,
    tip_labels=True,
    width=700,
    height=250,
)
a[0].label.text = "rooted tree"
a[1].label.text = "unrooted tree"
a[2].label.text = "re-rooted tree"
c
```

## The treenode

Every `ToyTree` has a `.treenode`, which is the top-level node in the data structure. That node exists whether the tree is rooted or unrooted. In an unrooted tree the treenode is not a biological root; it is just the top node in the current representation. Rooting inserts or relocates a node so that the treenode also becomes the biological root under the chosen interpretation.

When you root an unrooted tree, a new node is inserted on one edge. When you unroot a rooted tree, the current root split is collapsed. Re-rooting removes the current root and inserts a new one elsewhere.


```python
# the treenode is the top-level node
tree.treenode, tree[-1]
```

## Rooted vs unrooted drawings

A tree can be drawn with a linear rooted layout or with an unrooted layout, but those are separate choices from whether the tree is actually rooted in the data structure. The same topology can look rooted in one drawing and undirected in another.


```python
c, a, m = toytree.mtree([tree, utree, atree]).draw(
    layout="unrooted",
    node_labels="idx",
    node_sizes=14,
    tip_labels=True,
    width=700,
    height=260,
)
a[0].label.text = "rooted tree"
a[1].label.text = "unrooted tree"
a[2].label.text = "re-rooted tree"
c
```

## Which rooting method should I use?

- [Outgroup / manual rooting](mod-rooting-outgroup.md): use this when you know which tip or clade should be outside the ingroup.
- [Branch-length rooting](mod-rooting-branch-length.md): use this when you want to infer the root from edge lengths using midpoint, balanced midpoint, or MAD.
- [DLC rooting](mod-rooting-dlc.md): use this when rooting a gene tree against a rooted species tree with a gene-to-species mapping.

In general, manual outgroup rooting is preferred when the outgroup is known. Branch-length methods are useful when no outgroup is available. DLC rooting is specific to gene-tree / species-tree reconciliation problems.
